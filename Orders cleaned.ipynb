```python
product_dataset_cleaned.fillna({
    'product_category_name': 'unknown',
    'product_name_lenght': 0,
    'product_description_lenght': 0,
    'product_photos_qty': 0
}, inplace=True)
```

```python
numeric_cols = ['product_weight_g','product_length_cm','product_height_cm','product_width_cm']

for col in numeric_cols:
    product_dataset_cleaned.loc[:, col] = \
        product_dataset_cleaned[col].fillna(product_dataset_cleaned[col].median())
```

```python
product_dataset_cleaned.to_csv('product_dataset_cleaned.csv', index=False)
```
